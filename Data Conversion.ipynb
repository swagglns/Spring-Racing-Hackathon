```python
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd

file = 'converted_test_one_race.csv'
data = pd.read_csv(file)
```

```python
# race type

types = [
'Stakes',
'Claiming',
'Optional Claiming',
'Trial',
'Final',
'Maiden',
'Maiden Stakes',
'Maiden Claiming',
'Maiden Optional Claiming',
'Maiden Trial',
'Maiden Special Weight',
'Handicap',
'Handicap Stakes',
'Derby',
'Derby Trial',
'Derby Consolation',
'Speed Index Race',
'Speed Index Trial',
'Speed Index Final',
'Waiver Claiming',
'Waiver Maiden Claiming',
'Futurity',
'Futurity Trial',
'Futurity Consolation',
'Claiming Stakes',
'Claiming Stakes Trial',
'Invitational',
'Invitational Stakes',
'Allowance',
'Allowance Optional Claiming',
'Starter Allowance',
'Starter Optional Claiming',
'Starter Stakes',
'Starter Handicap',
'Stakes Consolation',
'Maturity Trial',
'Match Race',
'Championship'
]

for x in range(len(types)):
    type_mask = data['race_type'] == types[x]
    data.loc[type_mask, 'race_type'] = x

data.to_csv(file, index=False)
```

```python
# distance

yard_mask = data['distance_unit'] == 'Y'
data.loc[yard_mask, 'distance'] = data.loc[yard_mask, 'distance'] * 3
data.loc[yard_mask, 'distance_unit'] = 'F'

data.to_csv(file, index=False)
```

```python
# course

courses = [
'Hurdle',
'Dirt',
'Outer turf',
'Turf',
'Inner turf',
'Timber',
'All Weather Track',
'Downhill turf'
]

for x in range(len(courses)):
    type_mask = data['course'] == courses[x]
    data.loc[type_mask, 'course'] = x

data.to_csv(file, index=False)
```

```python
# surface

surfaces = [
'D',
'T'
]

for x in range(len(surfaces)):
    type_mask = data['surface'] == surfaces[x]
    data.loc[type_mask, 'surface'] = x

data.to_csv(file, index=False)
```

```python
# track condition

conditions = [
'GD',
'FT',
'MY',
'SY',
'FM',
'YL',
'HY',
'WF',
'SF',
'SL'
]

for x in range(len(conditions)):
    type_mask = data['track_condition'] == conditions[x]
    data.loc[type_mask, 'track_condition'] = x

data.to_csv(file, index=False)
```

```python
# weather

weathers = [
'Showery',
'Cloudy',
'Clear',
'Hazy',
'Snowing',
'Rainy',
'Foggy'
]

for x in range(len(weathers)):
    type_mask = data['weather'] == weathers[x]
    data.loc[type_mask, 'weather'] = x

data.to_csv(file, index=False)
```

```python
# sex

sexes = [
'Gelding',
'Ridgling',
'Mare',
'Horse',
'Filly',
'Colt',
'Spayed Mare'
]

for x in range(len(sexes)):
    type_mask = data['sex'] == sexes[x]
    data.loc[type_mask, 'sex'] = x

data.to_csv(file, index=False)
```

```python
# medication

meds = [
'L',
'BL',
'B',
'LA',
'A',
'CL',
'C',
'M'
]

for x in range(len(meds)):
    type_mask = data['medication'] == meds[x]
    data.loc[type_mask, 'medication'] = x

data.to_csv(file, index=False)
```
